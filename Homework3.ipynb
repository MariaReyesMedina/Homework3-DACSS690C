All the comparissons:

```python
import networkx as nx

G_learning = nx.from_pandas_adjacency(pairwise_learning, # just change this pairwase
                                 create_using=nx.DiGraph())

# pairwise
G_learning.edges(data=True)


import numpy as np
learning_comparisons ={(e[0],e[1]):e[2]['weight'] for e in G_learning.edges(data=True) if np.isfinite(e[2]['weight'])}

# ... and the remaining comparissons:
G_frinds = nx.from_pandas_adjacency(pairwise_frinds,create_using=nx.DiGraph())
frinds_comparisons={(e[0],e[1]):e[2]['weight'] for e in G_frinds.edges(data=True) if np.isfinite(e[2]['weight'])}

G_school_life = nx.from_pandas_adjacency(pairwise_school_life,create_using=nx.DiGraph())
school_life_comparisons={(e[0],e[1]):e[2]['weight'] for e in G_school_life.edges(data=True) if np.isfinite(e[2]['weight'])}

G_vocational_trng = nx.from_pandas_adjacency(pairwise_vocational_trng,create_using=nx.DiGraph())
vocational_trng_comparisons={(e[0],e[1]):e[2]['weight'] for e in G_vocational_trng.edges(data=True) if np.isfinite(e[2]['weight'])}

G_collage_prep = nx.from_pandas_adjacency(pairwise_collage_prep,create_using=nx.DiGraph())
collage_prep_comparisons={(e[0],e[1]):e[2]['weight'] for e in G_collage_prep.edges(data=True) if np.isfinite(e[2]['weight'])}

G_music_classes = nx.from_pandas_adjacency(pairwise_music_classes,create_using=nx.DiGraph())
music_classes_comparisons={(e[0],e[1]):e[2]['weight'] for e in G_music_classes.edges(data=True) if np.isfinite(e[2]['weight'])}

```

```python
# take a look
[learning_comparisons, frinds_comparisons,vocational_trng_comparisons,collage_prep_comparisons,music_classes_comparisons]
```

```python
# now the criteria

G_CRIT = nx.from_pandas_adjacency(pairwise_criteria,create_using=nx.DiGraph())
criteria_comparisons ={(e[0],e[1]):e[2]['weight'] for e in G_CRIT.edges(data=True) if np.isfinite(e[2]['weight'])}
criteria_comparisons
```

Now AHP

```python
!pip install AHPy
```

```python
import ahpy

learning = ahpy.Compare('learning', learning_comparisons, random_index='saaty')
frinds = ahpy.Compare('frinds', frinds_comparisons, random_index='saaty')
school_life = ahpy.Compare('school life', school_life_comparisons, random_index='saaty')
vocational_trng = ahpy.Compare('vocational trng', vocational_trng_comparisons, random_index='saaty')
collage_prep = ahpy.Compare('collage prep', collage_prep_comparisons, random_index='saaty')
music_classes = ahpy.Compare('music classes', music_classes_comparisons, random_index='saaty')
criteria = ahpy.Compare('criteria', criteria_comparisons, random_index='saaty')
```

Create hierarchy:


```python
criteria.add_children([learning, frinds, school_life, vocational_trng, collage_prep, music_classes])
```

Which criterion was more valuable: Learning

```python
weigths=criteria.global_weights
pd.Series(weigths).to_frame(name='Value')
```

Result:

```python
priorities=criteria.target_weights
pd.Series(priorities).to_frame(name='Value')
```

Assess consistency

```python
assesment=[(val.name,val.consistency_ratio) for val in [learning, frinds, school_life, vocational_trng, collage_prep, music_classes]]
pd.DataFrame(assesment,columns=['Criterion', 'Consistency Ratio'])
```

The vocational trng criterion has a consistency ratio of 0.20, which is above the 0.10 threshold. This indicates a relatively high level of inconsistency in the pairwise judgments, so these comparisons should be reviewed.
